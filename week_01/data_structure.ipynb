```python
def cal():
    n1 = float(input("Enter first number : "))
    n2 = float(input("Enter second number : "))
    op = input("Enter operator : ")
    if op == '+':
        return n1 + n2
    elif op == '-':
        return n1 - n2
    elif op == "*":
        return n1 * n2
    elif op == "/":
        if n2 == 0:
            print("denominaro must not be 0")
        else:
            return n1/n2
    else:
        print("invalid operator")
```

## variable length positional arguments

```python
def add(*values):     ## (10, 29, 54)
    total = 0
    for v in values:
        total += v       # total = total + v
        
    return total
```

```python
add(10, 29, 54)
```

```python
def print_details(**student_data):
    for key, value in student_data.items():
        print(f"{key:10s}     {value}")
```
