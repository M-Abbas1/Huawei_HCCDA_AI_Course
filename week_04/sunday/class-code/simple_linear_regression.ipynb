```python
x = np.array([1, 2, 3, 4, 5], dtype=float)   # hours studied  (input, x)
y = np.array([3, 5, 6.5, 9, 10.5])           # marks obtained (output, y)
```

```python
def f(x, w, b):
    y_hat = w*x + b
    return y_hat
```

```python
def plot_line(x, y, w, b):
    plt.scatter(x, y, c='red', marker='x', label="actual data")
    x_line = np.linspace(0, 6, 30)
    p = f(x_line, w, b)
    plt.plot(x_line, p, c='blue', lw=2, label=f"f(x) = {w}*X + {b}")
    plt.xlabel("Study Hours")
    plt.ylabel("obtained marks")
    plt.title("Study Hours vs obtained marks")
    plt.grid(alpha=0.4)
    plt.legend()
    plt.show()
```

```python
def cost(x, y, w, b):
    pred = f(x, w, b)
    error = y - pred
    c = sum(error**2)/(2 * len(x))
    return c
```

```python
def gradient_descent(x, y, w, b, alpha):
    y_hat = f(x, w, b)
    error = y_hat - y
    errorx = error*x

    w = w - alpha * (sum(errorx)/len(x))
    b = b - alpha * (sum(error)/len(x))

    return w, b
```

```python
w = 0
b = 0

epochs = 100
alpha = 0.1

costs = []

for i in range(epochs):
    w, b = gradient_descent(x, y, w, b, alpha)
    c = cost(x, y, w, b)

    costs.append(c)

    print(f"Epoch : {i+1}   New w : {w:.3f}  New b : {b:.3f}   cost : {c:.3f}")
```
