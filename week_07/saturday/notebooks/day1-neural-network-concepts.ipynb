### Demo: a neuron computing one output
Run this — it's a single neuron doing its job by hand. Inputs times weights, plus bias, through an activation. Nothing you haven't seen.

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# one neuron: 3 inputs, 3 weights, one bias
x = np.array([0.5, 0.8, 0.2])     # inputs
w = np.array([1.5, -2.0, 0.5])    # weights
b = 0.3                            # bias

z = np.dot(w, x) + b               # weighted sum + bias  (the "wx + b" step)
output = sigmoid(z)                # activation

print(f"weighted sum + bias:  z = {z:.3f}")
print(f"after sigmoid:        output = {output:.3f}")
print(f"\nThe output is between 0 and 1 — a probability, exactly like logistic regression.")
```

### Demo: the two activations from the slides
Run this to see sigmoid and ReLU — the shapes you saw on the slides, now in code.

```python
import matplotlib.pyplot as plt

def relu(z):
    return np.maximum(0, z)

z = np.linspace(-6, 6, 200)

fig, ax = plt.subplots(1, 2, figsize=(11, 3.5))
ax[0].plot(z, sigmoid(z), color="teal", linewidth=2.5)
ax[0].set_title("Sigmoid  →  squashes to 0–1")
ax[0].grid(alpha=0.3)
ax[1].plot(z, relu(z), color="orange", linewidth=2.5)
ax[1].set_title("ReLU  →  keeps positives, zeros the rest")
ax[1].grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
