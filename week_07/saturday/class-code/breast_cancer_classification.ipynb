```python
model = Sequential([
    Input(shape=(30,)),
    Dense(16, activation='relu'),
    Dense(8, activation='relu'),
    Dense(4, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    optimizer = "adam",
    loss= "binary_crossentropy",
    metrics = ["accuracy"]
)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=32,
    validation_split=0.2
)
```
