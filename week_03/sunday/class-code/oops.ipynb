```python
class Model:
    def __init__(self, m, b):
        self.m = m
        self.b = b

    def fit(self, x):
        result = []
        for el in x:
            y_hat = self.m*el + self.b
            result.append(y_hat)
        return result
```

```python
class Vehicle:
    def __init__(self, color, model, name):
        self.color = color
        self.model = model
        self.name = name

    def Break(self):
        print("__________________o")

    def print_details(self):
        print("we are in parent class function")
        print("*"*40)
        print(f"Car Name : {self.name}")
        print(f"Car Model : {self.model}")
        print(f"Color : {self.color}")
```

```python
class Car(Vehicle):
    def __init__(self, color="Black", name="", model=2000, dors=4, no_seat=4):
        super().__init__(color, model, name)
        self.dors = dors
        self.no_seat = no_seat

    def speed(self):
        print("almost 200k/h")

    def print_details(self):
        print("we are in child class function")
        super().print_details()
        print(f"NO doors : {self.dors}")
        print(f"NO Seats : {self.no_seat}")
        print("*"*40)
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self):
        dis = (self.x**2 + self.y**2)**(1/2)
        return dis

    def __str__(self):
        return f"P({self.x}, {self.y})"

    def __add__(self, other):
        tempx = self.x + other.x
        tempy = self.y + other.y
        return Point(tempx, tempy)
```

```python
p1 = Point(4, 3)
p2 = Point(9, 5)
```

```python
print(p1)
```

```python
from abc import ABC, abstractmethod
```

```python
class Shape(ABC):
    PI = 3.1415

    @abstractmethod
    def describe(self):
        print(f"show of type {type(self).__class}")


class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return Shape.PI*self.r**2

    def describe(self):
        print(f"Circle(radius={self.r})")
```

```python
c1 = Circle(10)
```

```python
c1.area()
```

```python
c1.describe()
```

```python
class FileHandler:
    def __init__(self):
        self.paths = None
        self.images = []

    def load_paths(self, file_path):
        with open(file_path, 'r') as f:
            self.paths = f.readlines()

    def store_image_names(self):
        for p in self.paths:
            self.images.append(p.split('/')[-1].strip())
```

```python
f1 = FileHandler()
```

```python
f1.images
```
