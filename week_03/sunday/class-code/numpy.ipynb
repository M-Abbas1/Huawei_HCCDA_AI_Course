```python
n = 1000000

my_list = list(range(n))

t0 = time.time()
list_result = [num*10 for num in my_list]
t1 = time.time()

total_time = t1-t0

print("total time : ", total_time*1000)
```

```python
numpy_arr = np.arange(n)

t0 = time.time()
arr_result = numpy_arr * 10
t1 = time.time()

total_time = t1 - t0



print("array time : ", total_time*1000)
```
