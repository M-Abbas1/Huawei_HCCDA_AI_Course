```python
balance = 1000


def transfer(amount):
    a = 10
    global balance
    if amount <= 0:
        raise ValueError(f"Can't transfer negative amount : {amount}")
    elif amount > balance:
        raise ValueError(f"Insufficient balance : your current balane is {balance}")
    else:
        balance -= amount

try:
    amount = float(input("Enter amount you wanna transfer : "))
    transfer(amount)
except ValueError as e:
    print(e)
        
```
